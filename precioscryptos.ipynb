```python
import requests

def obtener_precios_criptomonedas():
    url = 'https://api.coingecko.com/api/v3/simple/price?ids=bitcoin,ethereum,litecoin&vs_currencies=usd'
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        return data
    else:
        print("Error al obtener los precios de las criptomonedas.")
        return None

precios = obtener_precios_criptomonedas()

if precios:
    print("Precio de Bitcoin:", precios['bitcoin']['usd'])
    print("Precio de Ethereum:", precios['ethereum']['usd'])
    print("Precio de Litecoin:", precios['litecoin']['usd'])
```
